```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver
```

```python
llm = ChatGroq(model='openai/gpt-oss-20b')
search = GoogleSerperAPIWrapper()
```

```python
agent = create_agent(
    model=llm,
    tools=[search.run],
    system_prompt="You are a Agent and can Search for any Question in the Google Search Engine",
    checkpointer=MemorySaver()
)
```

```python
question="What is the Currnet International Run of Virat Kohli ?"
```

```python
response = agent.invoke({"messages":[{"role":"user","content":question}]},{"configurable": {"thread_id": "123"}})
```
