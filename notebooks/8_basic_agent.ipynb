```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
from langchain.tools import tool
from langchain_core.output_parsers import StrOutputParser

out = StrOutputParser()

load_dotenv()
```

```python
@tool
def add_number(a: int, b: int):
    """
       This Tool will return the Addition of Two Numbers 
    
       Args:
           a: First Number
           b: Second Number

    """
    return a + b

@tool
def multiply_number(a: int, b: int):
    """
       This Tool will return the Multiplication of Two Numbers 
    
       Args:
           a: First Number
           b: Second Number

    """
    return a * b
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")

agent = create_agent(
     model=llm,
    tools=[add_number,multiply_number],
    system_prompt="You are a Math Teacher, User tools for the Calculation"
)
```

```python
response = agent.invoke({"messages":[{"role":"user","content":"What is 2 sum 3 multiply by 10 ?"}]})
out.invoke(response['messages'][-1])
```
