```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(chunk_size=50,chunk_overlap=10)
```

```python
# Splitting the text
text = "My dear friends, students, and respected farmers,Today, I want to speak from the heart about three pillars that define the future of our nation: our youth, our education system, and our farmers.🎓 For the Youth You are the energy that drives our state forward."

text_chunk = splitter.split_text(text)
text_chunk
```

```python
# Splitting the Document

# Text Document Loader
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/speech.txt",encoding="utf-8")
texts = loader.load()
texts_chunks = splitter.split_documents(texts)
len(texts_chunks)
```
